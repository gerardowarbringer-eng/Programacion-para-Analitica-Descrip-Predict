### 1.1 Conversión (5 pts)

```python
# Escribe tu código aquí
#importamos datetime
import datetime as dt

# Muestra los datos origen
print("Tipos antes:")
print(taxis[['pickup', 'dropoff']].dtypes)

# Faltantes antes
faltantes_antes = taxis[['pickup', 'dropoff']].isna().sum()

# Formato de las fechas
formato = '%Y-%m-%d %H:%M:%S'

# Conversión
taxis['pickup'] = pd.to_datetime(taxis['pickup'], format=formato)
taxis['dropoff'] = pd.to_datetime(taxis['dropoff'], format=formato)

# Faltantes después
faltantes_despues = taxis[['pickup', 'dropoff']].isna().sum()

# Muestra los resultados
print("\nTipos después:")
print(taxis[['pickup', 'dropoff']].dtypes)

print("\nFaltantes antes:")
print(faltantes_antes)

print("\nFaltantes después:")
print(faltantes_despues)

print("\n¿Se generaron nuevos NaT?")
print(faltantes_despues > faltantes_antes)
```

## Actividad 2. Duraciones y anomalías (25 puntos)

### 2.1 Duración de cada viaje (6 pts)

1. Crea la columna `duracion` = `dropoff` − `pickup`.
2. Crea `duracion_min` con la duración en **minutos** (usa `.dt.total_seconds()`).
3. Muestra el resumen estadístico de `duracion_min` con `describe()`.

```python
# Escribe tu código aquí

# Crear columna duración
taxis['duracion'] = taxis['dropoff'] - taxis['pickup']

# Duración en minutos
taxis['duracion_min'] = taxis['duracion'].dt.total_seconds() / 60

# Resumen estadístico
print(taxis['duracion_min'].describe())
```

### 2.2 Duraciones anómalas (6 pts)

Cuenta los viajes con:

Muestra los viajes de duración cero con las columnas `pickup`, `dropoff`, `distance`, `fare` y `total`.

### 2.3 Velocidad promedio (6 pts)

La velocidad promedio de un viaje, en millas por hora, es `distance / (duracion_min / 60)`.

1. Crea la columna `velocidad_mph` **solo para viajes con duración mayor que cero**; en los demás debe quedar como faltante (`NaN`). Sugerencia: calcula la velocidad para todos los viajes y conserva solo los valores válidos con `.where(condición)`, que deja `NaN` donde la condición es `False`.
2. Cuenta los viajes con velocidad mayor a 60 mph (velocidad poco plausible en tráfico urbano).
3. Muestra los 5 viajes con mayor velocidad, ordenados de mayor a menor, con `pickup`, `dropoff`, `duracion_min`, `distance`, `velocidad_mph` y `fare`.

```python
# Escribe tu código aquí

# Calcula la velocidad y NaN cuando la duración sea 0 o negativa
taxis['velocidad_mph'] = (
    taxis['distance'] / (taxis['duracion_min'] / 60)
).where(taxis['duracion_min'] > 0)
print(taxis[['duracion_min', 'distance', 'velocidad_mph']].head())
```

```python
top_5_velocidad = (
    taxis.loc[
        taxis['velocidad_mph'] > 60,
        ['pickup', 'dropoff', 'duracion_min',
         'distance', 'velocidad_mph', 'fare']
    ]
    .nlargest(5, 'velocidad_mph')
)

print(top_5_velocidad)
```

## Actividad 3. Componentes y patrones (25 puntos)

### 3.1 Extraer componentes (6 pts)

Crea las columnas:

- `hora`: hora de `pickup`.
- `dia_num`: día de la semana de `pickup` (lunes = 0).
- `dia_nombre`: nombre del día de `pickup`.
- `fin_semana`: `True` si es sábado o domingo.

Cuenta cuántos viajes cruzan la medianoche.

```python
# Escribe tu código aquí
# Hora de pickup
taxis['hora'] = taxis['pickup'].dt.hour

# Día de la semana: lunes = 0, domingo = 6
taxis['dia_num'] = taxis['pickup'].dt.dayofweek

# Nombre del día
taxis['dia_nombre'] = taxis['pickup'].dt.day_name()

# Fin de semana: sábado (5) o domingo (6)
taxis['fin_semana'] = taxis['dia_num'].isin([5, 6])

# Solo la fecha, sin hora
taxis['fecha'] = taxis['pickup'].dt.date

# True si pickup y dropoff ocurren en fechas diferentes
taxis['cruza_medianoche'] = (
    taxis['pickup'].dt.date != taxis['dropoff'].dt.date
)
print(taxis[
    ['pickup', 'dropoff', 'hora', 'dia_num', 'dia_nombre',
     'fin_semana', 'fecha', 'cruza_medianoche']
].head())

```

### 3.2 Patrón por hora del día (7 pts)

Con `groupby('hora')`, calcula para cada hora: el **número de viajes** y la **mediana** de `duracion_min`. Indica la hora con más viajes y la hora con menos.

Usa la mediana (no la media) porque la duración tiene valores extremos.

```python
# Escribe tu código aquí
#Crea resumen por hora, usando groupby
resumen_hora = taxis.groupby('hora').agg(
    viajes=('duracion_min', 'count'),
    mediana_duracion_min=('duracion_min', 'median')
)

print(resumen_hora)
```

```python
# Calcula la hora con más y menos viajes
max_viajes = resumen_hora['viajes'].max()
min_viajes = resumen_hora['viajes'].min()

hora_max = resumen_hora[resumen_hora['viajes'] == max_viajes].index[0]
hora_min = resumen_hora[resumen_hora['viajes'] == min_viajes].index[0]

print(f"Hora con más viajes: {hora_max}")
print(f"Número de viajes: {max_viajes}")

print(f"Hora con menos viajes: {hora_min}")
print(f"Número de viajes: {min_viajes}")
```

### 3.3 Patrón por día de la semana (6 pts)

Calcula por día de la semana el **número de viajes** y la **mediana** de `duracion_min` y de `fare`. Presenta los días en orden de lunes a domingo (sugerencia: agrupa por `['dia_num', 'dia_nombre']`).

Toma en cuenta que marzo de 2019 no tiene el mismo número de cada día de la semana: calcula también el **promedio de viajes por fecha**, dividiendo entre el número de fechas distintas de ese día de la semana (usa la columna `fecha` y `.nunique()`).

```python
# Escribe tu código aquí
resumen_dia = (
    taxis.groupby(['dia_num', 'dia_nombre'])
    .agg(
        viajes=('duracion_min', 'count'),
        fechas_distintas=('fecha', 'nunique'),
        mediana_duracion_min=('duracion_min', 'median'),
        mediana_fare=('fare', 'median')
    )
    .sort_index()
)

# Promedio de viajes por fecha
resumen_dia['promedio_viajes_por_fecha'] = (
    resumen_dia['viajes'] / resumen_dia['fechas_distintas']
)

print(resumen_dia)
```

### 4.1 Dataset limpio (6 pts)

Crea `taxis_limpio` aplicando el criterio que definiste en 2.4. Imprime cuántos viajes se excluyeron por cada condición y en total, y verifica que `taxis_limpio` cumple todas las condiciones.

```python
# Escribe tu código aquí
# Condiciones de exclusión según el criterio de 2.4
cond_negativa = taxis['duracion_min'] < 0
cond_cero = taxis['duracion_min'] == 0
cond_larga = taxis['duracion_min'] > 180
cond_velocidad = taxis['velocidad_mph'] > 60

# Máscara final
cond_excluir = (
    cond_negativa |
    cond_cero |
    cond_larga |
    cond_velocidad
)

# Crear dataset limpio
taxis_limpio = taxis.loc[~cond_excluir].copy()

# Conteo por condición
print("Excluidos por duración negativa:", cond_negativa.sum())
print("Excluidos por duración cero:", cond_cero.sum())
print("Excluidos por duración > 3 horas:", cond_larga.sum())
print("Excluidos por velocidad > 60 mph:", cond_velocidad.sum())

# Total de filas realmente excluidas
print("Total excluidos:", cond_excluir.sum())

print("\nViajes originales:", len(taxis))
print("Viajes en taxis_limpio:", len(taxis_limpio))
```
